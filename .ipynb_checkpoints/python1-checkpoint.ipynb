```python
def prod(lst):
    s = 1
    for i in lst:
        s = s*i
    return s
```

```python
a = [2,3,4]
g = prod(a)
print(g)
```

```python
def cal(a,b):
    return (a+b, a-b, a*b, a/b)
```

```python
num = cal(5,2)
print(num)
```

ARIBATARY AGR - receives any number of argument

```python
def random_fun(*args):
    print(args)
    print(type(args))
    for i in args:
        print(i)
```

```python
number1 = (1,2,3)
result = random_fun(number1)
```
